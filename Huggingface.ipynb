```python
from transformers import pipeline
```

```python
from huggingface_hub.inference._generated.types import text_classification
# create the Pipeline(sequence of the different stages)
text_classification_pipeline = pipeline('text-classification',device=0)
```

```python
output = text_classification_pipeline('I love you')
output
```

```python
# Downdload the custom model for text classification
text_classification_pipeline = pipeline("text-classification",
                                        model="bhadresh-savani/distilbert-base-uncased-emotion",
                                        device=0)
```

```python
output = text_classification_pipeline("i love you")
output
```

```python
output = text_classification_pipeline("i hate you")
output
```

```python
output = text_classification_pipeline("india lost the cricket match")
output
```

```python
# NER
ner_pipeline = pipeline('ner',device=0)
```

```python
output = ner_pipeline("Steve job started Apple in 1976")
print(output)
```

```python
# text to speech
tts_pipeline = pipeline('text-to-speech')
```

```python
output = tts_pipeline("Differing views, observations are normal part of deliberation: ECI on reports of internal friction")
```

```python
import scipy

scipy.io.wavfile.write("output.wav",rate=16000,data=output["audio"])
```

```python
from bs4 import BeautifulSoup

#create a soup
soup = BeautifulSoup(response.text,"html.parser")
```

```python
# Get text from all paragraph tags on the page
paragraphs = soup.find_all("p")
content = "\n".join([p.get_text(strip=True) for p in paragraphs])

print(content)
```
